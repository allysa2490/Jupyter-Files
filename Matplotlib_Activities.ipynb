## Activity 1

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visits = [120, 150, 170, 160, 200, 220, 180]

fig, ax = plt.subplots()

ax.plot(days, visits, marker="o")
ax.set_title("Website Visits Over One Week")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Number of Visits")
ax.grid(True)

plt.show()
```

1.  **Which day has the highest visits? Which day has the lowest?** <br>
        *Saturday has the highest number of visits with 220, while Monday has the lowest with 120 visits.* <br><br>
2.  **Describe the overall pattern (increasing, decreasing, fluctuating).** <br>
        *The overall pattern of the data shows a generally increasing trend throughout the week with slight fluctuations. The number of visits rises from Monday to Wednesday, slightly decreases on Thursday, increases again toward Saturday, and then slightly drops on Sunday. Overall, website traffic is higher toward the weekend.* <br><br>
3.  **If you were a marketing analyst, what action would you recommend based on the trend?** <br>
        *If I were a marketing analyst, I would recommend increasing promotional activities at the beginning of the week, especially on Monday, to boost traffic during the lowest day. I would also maximize marketing efforts before the weekend since visits naturally increase during that time.* <br><br>

## Activity 2

```python
import matplotlib.pyplot as plt
products = ["A", "B", "C", "D", "E"]

sales = [32000, 28000, 41000, 15000, 36000]

# Sort from highest to lowest
pairs = sorted(zip(products, sales), key=lambda x: x[1], reverse=True)

products_sorted, sales_sorted = zip(*pairs)
fig, ax = plt.subplots()

ax.bar(products_sorted, sales_sorted)
ax.set_title("Total Sales by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Sales (PHP)")
ax.grid(True, axis="y")

plt.show()
```

1.  **Which product performs best? Which performs worst?** <br>
        *Product C performs the best with sales of 41,000 PHP, while Product D performs the worst with sales of 15,000 PHP.* <br><br>
2.  **Why are horizontal grid lines useful in bar charts?** <br>
        *Horizontal grid lines are useful because they make it easier to compare the heights of the bars and accurately estimate the sales values. They help viewers quickly see differences between products.* <br><br>
3.  **What business decision could be made using this plot?** <br>
        *Based on the plot, the business could focus on promoting and restocking high-performing products like Product C, while reviewing marketing strategies or improving Product D to increase its sales. The company may also allocate more resources to products that generate higher revenue.* <br><br>

## Activity 3

```python
import matplotlib.pyplot as plt

delivery_time = [22, 25, 23, 28, 35, 40, 24, 26, 27, 80, 90, 33, 30, 29, 25, 24]
fig, ax = plt.subplots()

ax.hist(delivery_time, bins=6)

ax.set_title("Distribution of Delivery Time")
ax.set_xlabel("Minutes")
ax.set_ylabel("Frequency")
ax.grid(True, axis="y")

plt.show()
```

1.  **What is the most common delivery time range?** <br>
        *The most common delivery time range is approximately 22 to 30 minutes, since most of the recorded delivery times fall within this interval.* <br><br>
2.  **Are there extreme delays? What values suggest outliers?** <br>
        *Yes, there are extreme delays. The values 80 and 90 minutes suggest outliers because they are much higher than the majority of delivery times, which are mostly below 40 minutes.* <br><br>
3.  **Give two operational reasons why delays might happen.** <br>
        *Two possible operational reasons for delays are heavy traffic conditions and a high volume of customer orders. Other factors such as vehicle breakdowns or weather disturbances could also contribute to longer delivery times.* <br><br>

## Activity 4

```python
import matplotlib.pyplot as plt

ad_spend = [5, 10, 10, 15, 20, 25, 30, 35, 40]
sales = [50, 60, 58, 75, 78, 88, 95, 96, 120]

fig, ax = plt.subplots()

ax.scatter(ad_spend, sales)
ax.set_title("Ad Spend vs Sales")
ax.set_xlabel("Ad Spend (PHP thousands)")
ax.set_ylabel("Sales (PHP thousands)")

ax.grid(True)
max_i = sales.index(max(sales))
ax.annotate("Highest sales", (ad_spend[max_i], sales[max_i]))

plt.show()
```

1.  **Do the points suggest a positive relationship? Explain briefly.** <br>
        *Yes, the points suggest a positive relationship because as ad spend increases, sales generally increase. This indicates that higher advertising budgets are associated with higher sales.* <br><br>
2.  **Is there an outlier point? What could cause it?** <br>
        *Yes, there is a possible outlier around an ad spend of 10 PHP thousands with sales of 58 PHP thousands. This could be caused by factors such as a special promotion, seasonal demand, or errors in data recording.* <br><br>
3.  **Why does correlation not automatically mean causation?** <br>
        *Correlation does not automatically mean causation because two variables can move together without one causing the other. Other factors, known as confounding variables, might influence the results, or the relationship could be coincidental.* <br><br>
